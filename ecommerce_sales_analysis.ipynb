```python
import pandas as pd

# sample ecommerce dataset
data = {
    'Order Date': ['2023-01-01','2023-01-05','2023-02-01','2023-02-10','2023-03-15'],
    'Product': ['Mobile','Shirt','Laptop','Shoes','Watch'],
    'Category': ['Electronics','Clothing','Electronics','Footwear','Accessories'],
    'Price': [10000,500,50000,2000,3000],
    'Quantity': [2,3,1,2,1]
}

df = pd.DataFrame(data)
df.head()
```

```python
df.info()
```

```python
df['Total Sales'] = df['Price'] * df['Quantity']
df.head()
```

```python
df.groupby('Product')['Total Sales'].sum()
```

```python
df.groupby('Category')['Total Sales'].sum()
```

```python
import matplotlib.pyplot as plt

df.groupby('Product')['Total Sales'].sum().plot(kind='bar')
plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Month'] = df['Order Date'].dt.month

df.head()
```

```python
monthly_sales = df.groupby('Month')['Total Sales'].sum()

monthly_sales
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array(monthly_sales.index).reshape(-1,1)
y = monthly_sales.values

model = LinearRegression()
model.fit(X, y)
```

```python
pred = model.predict([[4]])
print("Next month sales prediction:", pred)
```
